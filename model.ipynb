```python
import os
import cv2
import numpy as np



dataset_path = "ColorClassification" #Path where all photo exist

x=[]
y=[]

for color in os.listdir(dataset_path): # to get all Directory list
    folder_path = os.path.join(dataset_path, color) # joins dataset path and he color(basically the inside folders name ex:black,blue...)

    for image_file in os.scandir(folder_path): # this loop used to get all photos path
        img_To_Num = cv2.imread(image_file.path)    # reads the every single photo for every iterator
        if img_To_Num is None:                      #edge case , what if there is no photo , empty kind of file
            continue

        img_To_Num = cv2.resize(img_To_Num, (64, 64)) #Important step , will reduce the image to 64x64 , because each photos have different size/pixels
        img_To_Num = img_To_Num.flatten() # it makes 64x64x3

        x.append(img_To_Num)
        y.append(color)

print(len(x),len(y))
```

```python
from sklearn.model_selection import train_test_split

#splits into two part: train dataset and test dataset
x_train,x_test,y_train,y_test = train_test_split(x,y , test_size=0.2,random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
import joblib

#Specifying algorithm , in this case it is Random Forest
model = RandomForestClassifier(n_estimators=100,random_state=42)

#trains the model
model.fit(x_train,y_train)

#This Stores the trained data in the form of pickle file 
joblib.dump(model, "ColorClassifier.pkl")  #Important :
print("Model saved successfully!")
```

```python
from sklearn.metrics import accuracy_score,classification_report
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
#Testing the model by giving test dataset
predictions = model.predict(x_test)

#checking accuracy
accuracy = accuracy_score(y_test,predictions)

print(f"Model Accuracy : {accuracy}")
print(classification_report(y_test,predictions))

cm = confusion_matrix(y_test, predictions)

plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d')
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

After Training the model can run this below Cell anytime for get output for any input

```python
import cv2
import joblib

model = joblib.load("ColorClassifier.pkl")


def newInput(imgPath):
    img_To_Num = cv2.imread(imgPath)
    if img_To_Num is None:
        return "Invalid Input"
    img_To_Num = cv2.resize(img_To_Num, (64, 64))
    img_To_Num = img_To_Num.flatten()
    color_predictions = model.predict(img_To_Num.reshape(1,-1))
    return color_predictions


IMG_PATH = "For_Testing\\86.jpg" #give the image path which you want to predict the color
print(newInput(IMG_PATH))
```
